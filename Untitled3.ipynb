```python
import pandas as pd
df = pd.read_csv('Walmart_Sales.csv')
df.head()
```

```python
print(df.isnull().sum())
print(f"\nShape of data:{df.shape}")
```

```python
df['Date'] = pd.to_datetime(df['Date'] , dayfirst = True)
print(df['Date'].dt.year.unique())
```

```python
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.strftime('%W').astype(int)
df['Year'] = df['Date'].dt.year
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x= 'Temperature',y = 'Weekly_Sales' , data= df)
plt.title('Relationships between Temperature and Sales')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error , r2_score

features = ['Store','Holiday_Flag','Temperature',
           'Fuel_Price','CPI','Unemployment',
           'Month','Week','Year']

X = df[features]
Y = df['Weekly_Sales']
X_train, X_test , Y_train, Y_test = train_test_split(X,Y, test_size =0.2, random_state =42 )

model = RandomForestRegressor(n_estimators = 100, random_state= 42)
model.fit(X_train, Y_train)

Y_pred = model.predict(X_test)
print("MAE:", mean_absolute_error(Y_test,Y_pred))
print("R2 Score:", r2_score(Y_test, Y_pred))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

importance = model.feature_importances_

features = ['Store','Holiday_Flag','Temperature',
           'Fuel_Price','CPI','Unemployment',
           'Month','Week','Year']

feat_df = pd.DataFrame({'Feature': features, 'Importance': importance})
feat_df = feat_df.sort_values('Importance', ascending= True)

plt.figure(figsize=(10, 6))
plt.barh(feat_df['Feature'],feat_df['Importance'])
plt.title('Feature Importance')
plt.show()
```
